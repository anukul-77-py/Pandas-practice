```python
import pandas as pd
```

```python
dff={
    "Name":["Anukul",'Shyam','Hari','Sita'],
    'Age':[20,30,34,23],
    'Salary':[20000,25000,30000,32000]
}
```

```python
df=pd.DataFrame(dff)
df
```

```python
df['doj']=['15-1-2024','16-1-2024','17-3-2024','18-3-2024']
```

```python
df['doj']=pd.to_datetime(df['doj'],format='%d-%m-%Y')
df
```

```python
df['month']=df['doj'].dt.month
df
```

```python
#Aggreation and groupby
```

```python
df['month'].value_counts()# it gives frequency of values
```

```python
df[df['month']==1].value_counts()
```

```python
df.groupby('month')['Salary'].sum()#Aggregration based on group by
#it gives sum of same month
```

```python
df.groupby('month')['Salary'].mean()
```
