```python
dff={
    "Name":["ram",'hari','shyam'],
    'salary':[60000,70000,80000],
    'age':[20,21,22]
    
}
```

```python
df=pd.DataFrame(dff)
df
```

```python
df['doj']=['2026-8-9' , '2026-8-7', '2026-8-10']
df
```

```python
df['doj']=pd.to_datetime(df['doj'])
df['doj'].dtype
```
