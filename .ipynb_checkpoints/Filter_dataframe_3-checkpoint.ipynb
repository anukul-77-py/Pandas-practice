```python
import pandas as pd
```

```python
df=pd.read_csv('students.csv')
```

```python
print(df.head())
```

```python
(df['Computer']>80)#in this it give all value false and true but i want only true value in this case we do
```

```python
df[df['Computer']>80]#now i get all true value which is above 80
```

```python
print(df[(df['Computer']>80) & (df['Social']>80)])
```

```python
df.where((df['Computer']>80) & (df['Social']>80),other='not eligible').head()#where doesnt change the row and column of a table
```
