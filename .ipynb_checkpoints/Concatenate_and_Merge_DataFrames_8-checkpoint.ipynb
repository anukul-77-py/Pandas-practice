```python
import pandas as pd
```

```python
df=pd.DataFrame({
    'id':[1,2,3],
    'salary':[1000,2000,3000]
})
df
```

```python
df1=pd.DataFrame({
    'id':[3,4,5],
    'score':[10,11,12]
})
df1
```

```python
# Concatenate
```

```python
pd.concat([df,df1],axis=0,ignore_index=True)#it concat vertically , top on top,row level
```

```python
pd.concat([df,df1],axis=1)#horizontal/column level/ side on side
```

```python
#merge - joins 

```

```python
pd.merge(df,df1,how='inner',on='id')
```

```python
pd.merge(df,df1,how='outer',on='id')
```

```python
pd.merge(df,df1,how='inner',left_on='id',right_on='id')
```
